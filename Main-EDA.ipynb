# **Project Summary -**

Most workplace mental health analyses ask a single question: *how many people are unwell?* This project asks a more useful one: **where along the path from "I need help" to "I got help" do people actually drop out?**

The data is the **Mental Health in Tech Survey (2014)** from *Open Sourcing Mental Illness (OSMI)* — 1,259 technology workers across 48 countries, answering 27 questions about their demographics, their workplace, and their experience of mental health at work.

The analysis is built around a **four-stage funnel**, and every chart in this notebook is positioned against one of these stages:

| Stage | Question | What it measures |
|---|---|---|
| **1. Need** | Do I have a condition affecting me? | `work_interfere`, `family_history` |
| **2. Awareness** | Do I know what help exists? | `benefits`, `care_options`, `seek_help`, `wellness_program` |
| **3. Safety** | Is it safe for me to use it? | `anonymity`, `leave`, `obs_consequence`, the stigma questions |
| **4. Action** | Did I actually seek treatment? | `treatment` — the target |

Framing it this way changes what the numbers mean. A low treatment rate is not one problem; it is a sum of leaks at four different valves, and each valve has a different fix. The funnel chart (Chart 8) quantifies each stage directly, and it shows the biggest single drop is not at Need and not at Action — it is at **Awareness and Safety**, the two stages an employer controls most directly.

**Getting there required real cleaning.** `Age` contained a minimum of −1,726 and a maximum of 99,999,999,999 — a free-text box with no validation. `Gender` contained **49 distinct spellings** of three underlying categories, from case variants (`Male`, `m`, `M`) through typos (`Malr`, `msle`, `Femake`) to genuine non-binary self-descriptions (`Enby`, `Androgyne`, `fluid`). Three columns carried heavy missingness, and the central wrangling judgement was separating *structural* gaps (a blank `work_interfere` means "no condition" — informative, must be preserved) from *random* ones (`self_employed`, 18 rows, safely imputed). **Zero rows were dropped**: every fix repaired a cell rather than deleting a respondent.

**The analysis follows the UBM rule across 22 charts**, deliberately using a different visual grammar from the standard bar-and-pie approach — funnel charts, dumbbell plots for paired comparisons, slope charts for ordered categories, ridgelines for distribution overlap, a waffle chart for the target, and a tornado chart ranking every driver by effect size.

**What the funnel reveals.** Need is widespread: 62% report their condition interferes with work at least sometimes. But only **35% know their care options** and only **30% are confident their anonymity is protected** — and those two stages are where the funnel narrows hardest. Crucially, the data shows that *uncertainty behaves like absence*: respondents who did not know whether they had benefits sought treatment at **37.1%**, which is **lower** than those who knew they had none (48.2%). Not knowing is worse than knowing it's bad.

The strongest drivers of treatment are family history (74.0% vs 35.4%) and self-reported interference (85.0% vs 14.2%) — neither of which an employer can change. But **care-option awareness delivers a 27.7-point spread**, roughly three-quarters the effect size of family history, and it is fixable with a document rather than a budget. That comparison is the commercial heart of the project.

The recommendations that follow are ordered by cost-to-impact ratio, and the measurement section defines three trackable KPIs so the client can tell whether any of it worked.

# **Problem Statement**


**Write Problem Statement Here.**

A technology company can spend heavily on mental health benefits and still see almost no one use them. The spend is visible on the P&L; the reason for non-use is not visible anywhere.

This is a **funnel leakage problem**, and it is normally invisible because organisations measure only the two ends — money in, utilisation out — with no visibility into the three stages between them. An employee who never learns the benefit exists, and an employee who knows but fears their manager will find out, produce the identical utilisation statistic: zero. They require completely different interventions.

Using the OSMI 2014 survey, this project sets out to:

1. **Reconstruct the funnel** from the survey responses and quantify the population surviving each stage — Need, Awareness, Safety, Action.
2. **Locate the largest leak.** Which stage loses the most people?
3. **Separate controllable from uncontrollable drivers.** Rank every variable by its association with treatment-seeking, and mark which ones an employer can actually move.
4. **Measure the stigma component directly**, exploiting the survey's matched mental/physical question pairs as a built-in control.
5. **Convert the findings into interventions ranked by cost-to-impact**, with defined KPIs so the client can measure whether anything changed.

The deliverable is a prioritised action list where each recommendation names the funnel stage it repairs and the evidence supporting it.

#### **Define Your Business Objective?**

**Answer:**

The client is the **People Operations leadership of a technology company** that already spends money on mental health provision and cannot tell whether it is working.

The objective is to **raise the share of employees who get appropriate care, and cut the productivity loss from untreated conditions, at the lowest incremental spend.** Three sub-objectives follow directly from the funnel framing:

**A. Find the widest leak before spending anything.** Every stage has a different fix and they are not interchangeable. An awareness leak is fixed with a document. A safety leak is fixed with a policy and a systems audit. A provision leak is the only one that costs real money. Spending on the wrong stage produces a visible initiative with flat utilisation — the worst possible outcome, because it makes the *next* funding request harder.

**B. Rank the levers by cost-to-impact, not by impact alone.** Family history predicts treatment powerfully and cannot be changed. Care-option awareness predicts it almost as powerfully and can be changed this quarter for the price of writing something down. The ratio is what matters for budget allocation, and this analysis is designed to produce it.

**C. Establish a measurable baseline.** Wellbeing programmes are notoriously hard to evaluate, which is why they get cut first in a downturn. This analysis defines three concrete, re-surveyable metrics — mean support score, share uncertain about anonymity, and observed-consequence incidence — so that next year's report can show movement rather than sentiment.

The business return comes through earlier intervention (cheaper than late intervention), lower regrettable attrition in the expensive mid-career band, and reduced presenteeism among the ~62% reporting some work interference.

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# ==============================================================================
# Write your code to make your dataset analysis ready.
# ==============================================================================
# Seven idempotent steps on `df`. `raw` is never touched, so the final
# before/after comparison is genuine. Guiding principle throughout: REPAIR THE
# CELL, NEVER DELETE THE RESPONDENT.

print(f'Starting shape: {df.shape}')
print('=' * 68)

# ------------------------------------------------------------------------------
# [1] Timestamp -> datetime, plus derived calendar parts
# ------------------------------------------------------------------------------
# errors='coerce' converts unparseable strings to NaT rather than raising,
# which keeps the notebook runnable on a dirtier version of the same file.
df['Timestamp']    = pd.to_datetime(df['Timestamp'], errors='coerce')
df['survey_year']  = df['Timestamp'].dt.year
df['survey_month'] = df['Timestamp'].dt.month
print(f'[1] Timestamp parsed  |  {df["Timestamp"].min():%b %Y} → {df["Timestamp"].max():%b %Y}')

# ------------------------------------------------------------------------------
# [2] Age repair
# ------------------------------------------------------------------------------
# Only the offending CELL is nulled, never the row - each affected respondent
# still contributes 26 valid answers. Median imputation is used because it is
# robust to precisely the extreme outliers that caused the problem (the raw
# mean is ~7.9e7 thanks to a single 99-billion entry).
LO, HI = 15, 80
bad_age = (df['Age'] < LO) | (df['Age'] > HI)
n_bad   = int(bad_age.sum())

df.loc[bad_age, 'Age'] = np.nan
age_med = df['Age'].median()
df['Age'] = df['Age'].fillna(age_med).astype(int)
print(f'[2] Age repaired      |  {n_bad} impossible values → median {age_med:.0f}  '
      f'|  now {df["Age"].min()}–{df["Age"].max()}')

# ------------------------------------------------------------------------------
# [3] Gender normalisation: 49 variants -> 3 categories
# ------------------------------------------------------------------------------
# Explicit keyword sets rather than a regex, so the mapping stays auditable and
# is trivial to extend. Anything matching neither set falls to 'Other'.
MALE_SET = {
    'male', 'm', 'male-ish', 'maile', 'mal', 'male (cis)', 'make', 'man',
    'msle', 'mail', 'malr', 'cis man', 'cis male', 'male ', 'guy (-ish) ^_^',
    'male leaning androgynous', 'something kinda male?',
    'ostensibly male, unsure what that really means',
}
FEMALE_SET = {
    'female', 'f', 'woman', 'femake', 'female ', 'cis female', 'femail',
    'cis-female/femme', 'female (cis)', 'femaile', 'femal',
}


def clean_gender(value):
    """
    Collapse a raw free-text gender string to Male / Female / Other.

    'Other' groups genuine non-binary identities together with unparseable junk.
    This is an explicit analytical compromise: at n≈14 (~1%) the group cannot
    support reliable sub-analysis, but deleting it would erase respondents from
    the study entirely. It is reported, never sub-analysed.
    """
    if pd.isna(value):
        return 'Other'
    v = str(value).strip().lower()
    if v in MALE_SET:
        return 'Male'
    if v in FEMALE_SET:
        return 'Female'
    return 'Other'


df['Gender_Clean'] = df['Gender'].apply(clean_gender)
print(f'[3] Gender normalised |  {df["Gender"].nunique()} variants → 3  '
      f'|  {df["Gender_Clean"].value_counts().to_dict()}')

# ------------------------------------------------------------------------------
# [4] Missing values, treated according to WHY they are missing
# ------------------------------------------------------------------------------
# This is the most consequential judgement in the pipeline. Two of these gaps
# are structural (the blank IS the answer) and one is random.

# 4a — random gap, 18 rows: impute with mode
mode_se = df['self_employed'].mode()[0]
df['self_employed'] = df['self_employed'].fillna(mode_se)

# 4b — STRUCTURAL: the question was conditioned on having a condition, so a
#      blank most likely signals the absence of one. Mode-imputing here would
#      inflate 'Sometimes' by ~57% and manufacture a correlation that is not
#      in the data. We preserve the information as an explicit level instead.
df['work_interfere'] = df['work_interfere'].fillna('Not Applicable')

# 4c — STRUCTURAL: only US residents were asked. Label it honestly.
df['state'] = df['state'].fillna('Not in US')

# 4d — optional free text, 87% empty: keep the signal (did they elaborate?),
#      drop the unusable text
df['has_comment'] = df['comments'].notna().astype(int)
df = df.drop(columns=['comments'])

print(f'[4] Nulls handled     |  self_employed→mode({mode_se}), '
      f'work_interfere→"Not Applicable", state→"Not in US", comments→flag')
print(f'                      |  remaining nulls: {df.isnull().sum().sum()}')

# ------------------------------------------------------------------------------
# [5] Enforce ordinal ordering
# ------------------------------------------------------------------------------
# Without this, matplotlib sorts alphabetically and '100-500' plots before
# '6-25', turning every ordinal chart into nonsense.
SIZE_ORDER  = ['1-5', '6-25', '26-100', '100-500', '500-1000', 'More than 1000']
LEAVE_ORDER = ['Very easy', 'Somewhat easy', "Don't know",
               'Somewhat difficult', 'Very difficult']
WI_ORDER    = ['Never', 'Rarely', 'Sometimes', 'Often', 'Not Applicable']
WI_SEVERITY = ['Never', 'Rarely', 'Sometimes', 'Often']   # excludes N/A

df['no_employees']   = pd.Categorical(df['no_employees'],   SIZE_ORDER,  ordered=True)
df['leave']          = pd.Categorical(df['leave'],          LEAVE_ORDER, ordered=True)
df['work_interfere'] = pd.Categorical(df['work_interfere'], WI_ORDER,    ordered=True)
print('[5] Ordinal orders    |  no_employees, leave, work_interfere')

# ------------------------------------------------------------------------------
# [6] Feature engineering
# ------------------------------------------------------------------------------
# 6a — career stages: far more actionable for an HR audience than raw age
df['Age_Group'] = pd.cut(
    df['Age'], bins=[14, 25, 35, 45, 60, 80],
    labels=['15-25', '26-35', '36-45', '46-60', '60+'])

# 6b — composite awareness/support index: collapses five correlated categorical
#      columns into one interpretable 0-5 ordinal metric that can be tracked
#      over time as a KPI
SUPPORT_COLS = ['benefits', 'care_options', 'wellness_program',
                'seek_help', 'anonymity']
df['support_score'] = sum((df[c] == 'Yes').astype(int) for c in SUPPORT_COLS)

# 6c — numeric target for correlation work
df['treatment_binary'] = (df['treatment'] == 'Yes').astype(int)

# 6d — 48 countries is unusable in most charts; 4 groups is not
TOP3 = ['United States', 'United Kingdom', 'Canada']
df['Country_Group'] = df['Country'].where(df['Country'].isin(TOP3), 'Other')

print('[6] Features built    |  Age_Group, support_score, treatment_binary, Country_Group')

# ------------------------------------------------------------------------------
# [7] Validate
# ------------------------------------------------------------------------------
print('=' * 68)
print(f'Before : {raw.shape}')
print(f'After  : {df.shape}')
print(f'Rows lost : {raw.shape[0] - df.shape[0]}   <- zero; no respondent discarded')
print(f'Nulls     : {df.isnull().sum().sum()}')
print(f'Added     : {[c for c in df.columns if c not in raw.columns]}')

# Module-level constants reused by the chart cells below
BASE_RATE = df['treatment_binary'].mean() * 100
N         = len(df)
print(f'\nBase treatment rate: {BASE_RATE:.1f}%  (n = {N})')

display(df.head())
````

### What all manipulations have you done and insights you found?

**Answer:**

**The seven manipulations**

**[1] Timestamp parsing.** Converted to true datetime with `errors='coerce'`, deriving `survey_year` and `survey_month`. Confirmed the collection window as Aug 2014 – Feb 2016, with volume overwhelmingly concentrated in the first days — the signature of a link shared on social media rather than a sampled survey.

**[2] Age repair.** Values outside 15–80 were nulled and imputed with the **median (31)**. Two choices worth defending: the median over the mean, because the raw mean was ~7.9 × 10⁷ years — destroyed by a single 99-billion entry; and **cell-level repair over row deletion**, so each of the five affected respondents kept their 26 other valid answers.

**[3] Gender normalisation.** **49 raw strings → 3 categories** via explicit keyword sets. Result: **990 Male (79.1%), 247 Female (19.7%), 14 Other (1.1%)**. The 'Other' bucket honestly combines genuine non-binary identities with junk entries — at n = 14 it is reported but never sub-analysed, since a two-person shift moves its percentage by 14 points.

**[4] Missing values, differentiated by cause.** The most consequential decision in the pipeline. `self_employed` (random, 18 rows) was mode-imputed. `work_interfere` (264 rows) was **not** — it is structurally missing, conditioned on having a condition, so a blank *is* a response. Mode-imputing it would have inflated "Sometimes" by roughly 57% and fabricated a correlation. `state` became `"Not in US"`; `comments` became a boolean flag.

**[5] Ordinal ordering** enforced on `no_employees`, `leave` and `work_interfere` via `pd.Categorical`, without which every ordinal chart sorts alphabetically and becomes meaningless.

**[6] Feature engineering.** `Age_Group` (career stages), `support_score` (0–5 composite index), `treatment_binary`, `Country_Group`.

**[7] Validation.** 1,259 rows in, 1,259 out, zero nulls remaining.

---

**Insights found during wrangling — before plotting anything**

**The data quality is itself a finding.** A form that accepted an age of −1,726 and 49 spellings of "male" had no input validation whatsoever. Any organisation replicating this survey should constrain inputs at capture — and the analyst should treat every free-text field as suspect by default.

**The 21% blank rate on `work_interfere` is a result, not a gap.** Roughly one in five respondents effectively self-reported having no mental health condition. This is arguably the most important single wrangling insight in the project: had it been mode-imputed, the entire severity analysis in Chart 14 would have been corrupted, and the dose-response relationship — the strongest internal validity check available — would have been destroyed.

**The target is balanced at 637/622 (50.6% / 49.4%).** Unusually convenient. No resampling or class weighting is required, and every subgroup rate can be read directly against a clean 50% line.

**Zero rows lost.** The pipeline is non-destructive at the row level throughout. That matters statistically, and it matters ethically for a survey about a stigmatised topic where every response was voluntarily given.

**`support_score` skews hard toward zero**, foreshadowing the central funnel finding: most respondents can confirm very few of the five supports — and, as Chart 8 will show, that is a knowledge problem far more than a provision one.

#### Chart - 5 — Severity of Work Interference

````python
# ==============================================================================
# Chart - 5 : work_interfere (Univariate, Ordinal)
# ==============================================================================
# A diverging horizontal bar, split at the "any interference" boundary. Placing
# 'Never' and 'Not Applicable' to the LEFT of zero and the three severity levels
# to the RIGHT lets the reader see the healthy/affected split as a single
# visual balance rather than by adding up categories mentally.

fig, axes = plt.subplots(1, 2, figsize=(15, 5.4))

# ---- Left: diverging severity bar ------------------------------------------
ax = axes[0]
counts = df['work_interfere'].value_counts().reindex(WI_ORDER)
pcts   = counts / N * 100

# Left of zero = no reported interference; right of zero = some interference
left_side  = [('Not Applicable', GREY), ('Never', TEAL)]
right_side = [('Rarely', '#F2C57C'), ('Sometimes', OCHRE), ('Often', FLAME)]

cursor = 0
for lab, c in left_side:
    w = pcts[lab]
    ax.barh(0, -w, left=cursor, color=c, edgecolor='white', height=0.5)
    ax.text(cursor - w / 2, 0, f'{lab}\n{w:.1f}%', ha='center', va='center',
            fontsize=9, color='white', fontweight='bold')
    cursor -= w

cursor = 0
for lab, c in right_side:
    w = pcts[lab]
    ax.barh(0, w, left=cursor, color=c, edgecolor='white', height=0.5)
    ax.text(cursor + w / 2, 0, f'{lab}\n{w:.1f}%', ha='center', va='center',
            fontsize=9, color='white' if lab != 'Rarely' else INK,
            fontweight='bold')
    cursor += w

ax.axvline(0, color=INK, lw=2.2)
ax.text(0, 0.42, 'no interference  ←  |  →  some interference',
        ha='center', fontsize=9.5, color=INK, fontweight='bold')
ax.set_ylim(-0.6, 0.62)
ax.set_yticks([])
ax.set_xlim(-45, 68)
finish(ax, 'Does your condition interfere with work?',
       f'Diverging at the any-interference boundary; n = {N}',
       xlabel='% of all respondents', despine=False)
ax.grid(axis='x', alpha=0.35)

# ---- Right: severity among those WITH a condition --------------------------
# Conditioning on having a condition is the clinically meaningful view; the
# large 'Not Applicable' block otherwise dominates and masks the shape.
ax = axes[1]
cond   = df[df['work_interfere'] != 'Not Applicable']
c_cnt  = cond['work_interfere'].value_counts().reindex(WI_SEVERITY)
c_pct  = c_cnt / len(cond) * 100
colors = [TEAL, '#F2C57C', OCHRE, FLAME]

bars = ax.bar(range(len(c_pct)), c_pct.values, color=colors,
              edgecolor='white', width=0.62)
for i, (v, n_) in enumerate(zip(c_pct.values, c_cnt.values)):
    ax.text(i, v + 1.2, f'{v:.1f}%\n(n={n_})', ha='center',
            fontsize=9.5, fontweight='bold', color=INK)
ax.set_xticks(range(len(c_pct)))
ax.set_xticklabels(WI_SEVERITY)
ax.set_ylim(0, max(c_pct.values) * 1.26)
finish(ax, 'Severity among respondents with a condition',
       f'Excludes "Not Applicable"; n = {len(cond)}',
       xlabel='', ylabel='% of affected respondents')

plt.tight_layout()
plt.show()

any_int = df['work_interfere'].isin(['Rarely', 'Sometimes', 'Often']).sum()
print(f'Any interference : {any_int} ({any_int/N*100:.1f}% of all respondents)')
print(f'Often            : {counts["Often"]} ({pcts["Often"]:.1f}%)')
print(f'Not Applicable   : {counts["Not Applicable"]} ({pcts["Not Applicable"]:.1f}%)')
````

##### 1. Why did you pick the specific chart?

**Answer:** A **diverging bar split at the any-interference boundary** is chosen over a standard ordinal bar chart because the primary question here is a *balance* question: what share of this workforce is unaffected versus affected? Placing `Never` and `Not Applicable` to the left of zero and the three severity levels to the right lets the reader see that split as a single visual weight, instead of mentally summing three categories.

The severity gradient is encoded **redundantly in colour** (teal → amber → vermillion), so escalation registers pre-attentively without reading labels.

The second panel **conditions on having a condition**, which is the clinically meaningful denominator. The 264-respondent `Not Applicable` block — the structural blanks preserved during wrangling — is large enough to dominate the unconditional view and flatten the severity distribution among the affected group, which is the population the client actually needs to plan for.

# **ACT II — THE FUNNEL**
*Charts 8–12. Still univariate, but organised around the question that structures this entire project: where between "I need help" and "I got help" do people actually drop out?*


#### Chart - 8 — The Help-Seeking Funnel

````python
# ==============================================================================
# Chart - 8 : THE CENTRAL CHART OF THIS PROJECT
# ==============================================================================
# A funnel quantifying survivors at each stage: Need -> Awareness -> Safety ->
# Action. Each stage is measured on the SAME denominator (all respondents) so
# the widths are directly comparable and the largest drop is visually obvious.

# ------------------------------------------------------------------------------
# Define each stage and the condition that counts a respondent as "surviving" it
# ------------------------------------------------------------------------------
stages = [
    ('All respondents',
     pd.Series(True, index=df.index),
     'Everyone surveyed'),
    ('1. NEED — reports a condition',
     df['work_interfere'] != 'Not Applicable',
     'Answered the conditional severity question'),
    ('2. AWARENESS — knows care options',
     df['care_options'] == 'Yes',
     'Can name what support is available'),
    ('3. SAFETY — confident of anonymity',
     df['anonymity'] == 'Yes',
     'Believes usage stays confidential'),
    ('4. ACTION — sought treatment',
     df['treatment'] == 'Yes',
     'Actually obtained care'),
]

labels  = [s[0] for s in stages]
counts  = [int(s[1].sum()) for s in stages]
subs    = [s[2] for s in stages]
pcts    = [c / N * 100 for c in counts]

fig, ax = plt.subplots(figsize=(13, 7))

stage_colors = [GREY, PLUM, OCHRE, FLAME, TEAL]

for i, (lab, cnt, pct, c) in enumerate(zip(labels, counts, pcts, stage_colors)):
    y = len(stages) - 1 - i
    # Centre each bar so the shape reads as a funnel narrowing downward
    ax.barh(y, pct, left=(100 - pct) / 2, color=c,
            edgecolor='white', height=0.62)
    ax.text(50, y, f'{pct:.1f}%   (n = {cnt})', ha='center', va='center',
            color='white', fontweight='bold', fontsize=11.5)
    ax.text(-2, y, lab, ha='right', va='center', fontsize=10.5,
            color=INK, fontweight='bold')
    ax.text(102, y, subs[i], ha='left', va='center',
            fontsize=8.5, color='#7A7A7A')

    # Annotate the drop between consecutive stages
    if i > 0:
        drop = pcts[i - 1] - pct
        ax.annotate(f'−{drop:.1f} pp', xy=(50, y + 0.5),
                    ha='center', fontsize=9.5, color=FLAME,
                    fontweight='bold')

ax.set_xlim(-30, 132)
ax.set_ylim(-0.6, len(stages) - 0.2)
ax.set_yticks([])
ax.set_xticks([])
ax.grid(False)
for spine in ax.spines.values():
    spine.set_visible(False)
ax.set_title('The help-seeking funnel: where do people drop out?',
             loc='left', fontsize=14, pad=26)
ax.text(-30, len(stages) - 0.35,
        'Each stage measured against the full sample, so widths are comparable',
        fontsize=9.5, color='#6B6B6B')

plt.tight_layout()
plt.show()

print('STAGE SURVIVORS')
print('=' * 58)
for lab, cnt, pct in zip(labels, counts, pcts):
    print(f'  {lab:<38} {cnt:>5}  {pct:5.1f}%')
print('\nLARGEST SINGLE DROP:')
drops = [(labels[i], pcts[i-1] - pcts[i]) for i in range(1, len(pcts))]
worst = max(drops, key=lambda t: t[1])
print(f'  into "{worst[0]}"  —  {worst[1]:.1f} percentage points')
````

##### 1. Why did you pick the specific chart?

**Answer:** This is the **organising chart of the entire project**, and the funnel form is chosen because it does something no bar chart can: it makes *sequential attrition* visible as a shape.

The critical design decision is that **every stage is measured against the same denominator — all 1,259 respondents — rather than against the preceding stage.** That makes the bar widths directly comparable and the largest narrowing visually obvious. Measuring each stage against the previous one (the conventional conversion-rate funnel) would have produced percentages that look reasonable in isolation while hiding the cumulative collapse.

Centring each bar produces the classic funnel silhouette, and annotating the **percentage-point drop between consecutive stages** puts the diagnostic number exactly where the reader's eye already is.

The stages themselves are constructed rather than given — each is a defensible proxy drawn from a survey question, and the construction is documented in the code so the reasoning is auditable.

* **Stage 4 — Action: 637 (50.6%)** actually sought treatment

**The funnel does not narrow where intuition expects.** Need is enormous — roughly four in five respondents. But the collapse happens immediately after: **the drop from Need into Awareness is over 43 percentage points**, by far the largest single leak in the sequence. #### Chart - 11 — The Composite Support Score

````python
# ==============================================================================
# Chart - 11 : engineered support_score (0-5)
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(15, 5.2))

# ---- Left: distribution of the index ---------------------------------------
ax = axes[0]
sc = df['support_score'].value_counts().sort_index()
grad = [FLAME, '#EE8959', OCHRE, '#B5C99A', '#7FC8A9', TEAL]

bars = ax.bar(sc.index, sc.values, color=[grad[i] for i in sc.index],
              edgecolor='white', width=0.68)
for i, v in zip(sc.index, sc.values):
    ax.text(i, v + 8, f'{v}\n{v/N*100:.1f}%', ha='center',
            fontsize=9.5, fontweight='bold', color=INK)

mean_sc = df['support_score'].mean()
ax.axvline(mean_sc, color=INK, ls='--', lw=2)
ax.text(mean_sc + 0.12, sc.max() * 0.88, f'mean\n{mean_sc:.2f}',
        fontsize=10, fontweight='bold', color=INK)

ax.set_xticks(range(6))
ax.set_ylim(0, sc.max() * 1.25)
finish(ax, 'How many supports can each respondent confirm?',
       'Composite of benefits, care options, wellness programme, seek-help, anonymity',
       xlabel='Support score (0–5)', ylabel='Respondents')

# ---- Right: cumulative view -------------------------------------------------
# The ECDF answers "what share can confirm AT MOST k supports?", which is the
# framing that makes the shortfall concrete
ax = axes[1]
cum = sc.cumsum() / N * 100
ax.step(sc.index, cum.values, where='post', color=PLUM, lw=2.8)
ax.fill_between(sc.index, 0, cum.values, step='post', color=PLUM, alpha=0.13)
ax.scatter(sc.index, cum.values, s=110, color=PLUM, zorder=3,
           edgecolor='white', linewidth=1.5)

for i, v in zip(sc.index, cum.values):
    ax.text(i, v + 3.5, f'{v:.0f}%', ha='center', fontsize=9.5,
            fontweight='bold', color=INK)

ax.axhline(50, color=FLAME, ls=':', lw=1.6)
ax.text(4.3, 52, 'half the sample', fontsize=9, color=FLAME)
ax.set_xticks(range(6))
ax.set_ylim(0, 108)
finish(ax, 'Cumulative share by support score',
       'Read as: X% can confirm AT MOST this many supports',
       xlabel='Support score (0–5)', ylabel='Cumulative % of respondents')

plt.tight_layout()
plt.show()

print(f"Mean   : {df['support_score'].mean():.2f} / 5")
print(f"Median : {df['support_score'].median():.0f} / 5")
print(f"Score 0   : {(df['support_score']==0).sum()} "
      f"({(df['support_score']==0).mean()*100:.1f}%)")
print(f"Score 0–1 : {(df['support_score']<=1).mean()*100:.1f}%")
print(f"Score 4–5 : {(df['support_score']>=4).mean()*100:.1f}%")
````

#### Chart - 14 — Severity and the Recognition Threshold

````python
# ==============================================================================
# Chart - 14 : work_interfere vs treatment (Bivariate, Ordinal-Cat)
# ==============================================================================
# A step/line plot across an ORDERED exposure scale. When an outcome rises
# monotonically with an ordered exposure, that is a DOSE-RESPONSE relationship -
# one of the strongest available signals that an association is real rather than
# coincidental. Only a connected line makes monotonicity visible; four separate
# bars would show four values but no trend.

fig, axes = plt.subplots(1, 2, figsize=(15, 5.3))

# ---- Left: the dose-response curve -----------------------------------------
ax = axes[0]
sev_rates = [df[df['work_interfere'] == s]['treatment_binary'].mean() * 100
             for s in WI_SEVERITY]
sev_n     = [int((df['work_interfere'] == s).sum()) for s in WI_SEVERITY]
x = np.arange(len(WI_SEVERITY))

ax.plot(x, sev_rates, color=FLAME, lw=3, zorder=2)
ax.fill_between(x, BASE_RATE, sev_rates, color=FLAME, alpha=0.10)
ax.scatter(x, sev_rates, s=240, color=FLAME, zorder=3,
           edgecolor='white', linewidth=2)

for i, (v, n_) in enumerate(zip(sev_rates, sev_n)):
    ax.text(i, v + 5, f'{v:.1f}%', ha='center', fontsize=11,
            fontweight='bold', color=INK)
    ax.text(i, v - 8, f'n={n_}', ha='center', fontsize=8.5, color='#7A7A7A')

ax.axhline(BASE_RATE, color=GREY, ls=':', lw=1.8)
ax.text(3.05, BASE_RATE + 2, f'base {BASE_RATE:.1f}%', fontsize=9,
        color='#7A7A7A', ha='right')

# Highlight the threshold jump - the actual finding of this chart
ax.annotate('', xy=(1, sev_rates[1] - 3), xytext=(0, sev_rates[0] + 3),
            arrowprops=dict(arrowstyle='->', color=PLUM, lw=2.4))
ax.text(0.5, (sev_rates[0] + sev_rates[1]) / 2 - 4,
        f'+{sev_rates[1]-sev_rates[0]:.1f} pp\nin ONE step',
        ha='center', fontsize=10, color=PLUM, fontweight='bold')

ax.set_xticks(x)
ax.set_xticklabels(WI_SEVERITY, fontsize=10)
ax.set_ylim(0, 100)
finish(ax, 'Treatment rate rises with reported severity',
       'Monotonic dose-response — strong evidence the relationship is real',
       xlabel='Reported work interference', ylabel='% who sought treatment')

# ---- Right: step sizes, to expose the threshold structure ------------------
# Plotting the DIFFERENCES rather than the levels makes the point unmissable:
# almost the entire effect happens in the first step.
ax = axes[1]
steps = [sev_rates[i + 1] - sev_rates[i] for i in range(len(sev_rates) - 1)]
step_labels = [f'{WI_SEVERITY[i]}\n→ {WI_SEVERITY[i+1]}'
               for i in range(len(sev_rates) - 1)]
colors = [PLUM if s == max(steps) else '#C9C9C9' for s in steps]

bars = ax.bar(range(len(steps)), steps, color=colors,
              edgecolor='white', width=0.55)
for i, v in enumerate(steps):
    ax.text(i, v + 1.6, f'+{v:.1f} pp', ha='center',
            fontsize=11, fontweight='bold', color=INK)

ax.set_xticks(range(len(steps)))
ax.set_xticklabels(step_labels, fontsize=9.5)
ax.set_ylim(0, max(steps) * 1.25)
finish(ax, 'Size of each step up the severity scale',
       'Nearly the whole effect occurs at the first threshold',
       xlabel='', ylabel='Change in treatment rate (pp)')

plt.tight_layout()
plt.show()

for s, r, n_ in zip(WI_SEVERITY, sev_rates, sev_n):
    print(f'  {s:<12} n={n_:<5} treatment rate = {r:5.1f}%')
print(f'\nNever → Rarely jump : +{sev_rates[1]-sev_rates[0]:.1f} pp')
print(f'Rarely → Often       : +{sev_rates[-1]-sev_rates[1]:.1f} pp')
````

Only a connected line makes monotonicity visible as a property; four separate bars would display four values but no trend.

`Not Applicable` is deliberately excluded, because it is not a severity level. Including it would break the ordinal scale and destroy the very pattern the chart exists to demonstrate.

#### Chart - 17 — Age: A Deliberate Null Result

````python
# ==============================================================================
# Chart - 17 : Age vs treatment (Bivariate, Numerical-Categorical)
# ==============================================================================
# RIDGELINE. When the question is "do these two groups differ?", the answer is
# about DISTRIBUTION OVERLAP - and overlap is what a ridgeline shows better than
# any summary statistic. Offsetting the two densities vertically keeps both
# shapes readable while the shared x-axis makes any horizontal shift obvious.

fig, axes = plt.subplots(1, 2, figsize=(15, 5.3),
                         gridspec_kw={'width_ratios': [1.25, 1]})

# ---- Left: ridgeline of age by treatment status ----------------------------
ax = axes[0]
from scipy.stats import gaussian_kde

grid = np.linspace(15, 70, 300)
offsets = {'No': 0.0, 'Yes': 0.55}
cols_   = {'No': GREY, 'Yes': TEAL}

for status in ['No', 'Yes']:
    vals = df[df['treatment'] == status]['Age'].values
    kde  = gaussian_kde(vals)
    dens = kde(grid)
    dens = dens / dens.max() * 0.48          # normalise height for comparability
    off  = offsets[status]

    ax.fill_between(grid, off, dens + off, color=cols_[status],
                    alpha=0.62, zorder=2 if status == 'Yes' else 1)
    ax.plot(grid, dens + off, color=INK, lw=1.4, zorder=3)

    med = np.median(vals)
    ax.plot([med, med], [off, off + 0.48], color=FLAME, lw=2, ls='--', zorder=4)
    ax.text(med + 0.8, off + 0.40, f'median {med:.0f}',
            fontsize=9, color=FLAME, fontweight='bold')
    ax.text(16, off + 0.30, f'Treatment: {status}\n(n={len(vals)})',
            fontsize=10, fontweight='bold', color=INK)

ax.set_yticks([])
ax.set_xlim(15, 70)
finish(ax, 'Age distribution by treatment status',
       'Near-total overlap — age does not separate the two groups',
       xlabel='Age (years)', ylabel='')

# ---- Right: treatment rate across career stages ----------------------------
ax = axes[1]
ag_rate = df.groupby('Age_Group', observed=True)['treatment_binary'].mean() * 100
ag_n    = df['Age_Group'].value_counts().reindex(ag_rate.index)

ax.plot(range(len(ag_rate)), ag_rate.values, color=PLUM, lw=2.6,
        marker='o', markersize=13, markeredgecolor='white', markeredgewidth=2)
ax.axhline(BASE_RATE, color=GREY, ls=':', lw=2)
ax.text(len(ag_rate) - 1, BASE_RATE + 2.2, f'base rate {BASE_RATE:.1f}%',
        fontsize=9, color='#7A7A7A', ha='right')

for i, (v, n_) in enumerate(zip(ag_rate.values, ag_n.values)):
    ax.text(i, v + 4.5, f'{v:.0f}%', ha='center', fontsize=10,
            fontweight='bold', color=INK)
    ax.text(i, v - 7, f'n={n_}', ha='center', fontsize=8, color='#7A7A7A')

ax.set_xticks(range(len(ag_rate)))
ax.set_xticklabels(ag_rate.index, fontsize=10)
ax.set_ylim(0, 100)
finish(ax, 'Treatment rate by career stage',
       'Flat within the well-populated bands',
       xlabel='', ylabel='% who sought treatment')

plt.tight_layout()
plt.show()

print(df.groupby('treatment')['Age'].agg(['count', 'mean', 'median', 'std']).round(2))
print('\nTreatment rate by career stage:')
for stage, r, n_ in zip(ag_rate.index, ag_rate.values, ag_n.values):
    print(f'  {str(stage):<8} n={n_:<5} {r:5.1f}%')
````

##### 1. Why did you pick the specific chart?

**Answer:** A **ridgeline** is the right encoding when the question is *"do these two groups differ?"*, because the honest answer to that question is about **distribution overlap** — and overlap is exactly what summary statistics conceal. Two groups can differ by a full year in the mean while being visually indistinguishable, which is precisely what happens here.

Offsetting the two densities vertically keeps both shapes fully readable (overlaid curves would obscure each other) while the shared x-axis makes any horizontal shift immediately obvious. Normalising each density to the same peak height is deliberate: it makes the *shapes* comparable rather than the group sizes, which is the comparison being asked for.

A boxplot was rejected — it would reduce each group to five numbers and hide precisely the overlap that constitutes the finding.

##### 2. What is/are the insight(s) found from the chart?

**Answer:**

* **Treated: mean 32.6, median 32.** **Untreated: mean 31.5, median 31.**
* The difference is roughly **1.1 years** — detectable in a sample this size, and practically meaningless.
* The two ridgelines are **almost perfectly superimposed**. Both are right-skewed, both peak near 30, both have thin tails past 45.
* Across career stages the rate hovers near the 50.6% base rate in every well-populated band.

**Interpretation — and this negative result is genuinely valuable.** **Age does not meaningfully predict treatment-seeking.** Given that family history separates the groups by 38.6 points and work interference by 70.8, the near-total absence of an age effect is informative rather than disappointing.

The mild upward drift in older bands is best read as an **accumulation effect** rather than a risk effect: the question asks whether the respondent has *ever* sought treatment, so a 45-year-old has simply had more years in which it could have happened. Adjusted for exposure time, even that gradient likely disappears.

Not every variable needs to produce a headline. Establishing that age is *not* a driver eliminates a whole class of alternative explanations and sharpens the case for the factors that are.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

**Answer:**

**Positive business impact.** The practical conclusion is that **support should not be age-segmented and no age band should be deprioritised.** This is useful because it closes off an intuitively appealing but unsupported allocation strategy — a client might reasonably have assumed younger employees need more support (more transition, less security) or that older employees do (more accumulated stress, more caregiving load). #### Chart - 20 — Ranking Every Driver by Effect Size

````python
# ==============================================================================
# Chart - 20 : Tornado chart of effect sizes (Multivariate summary)
# ==============================================================================
# A TORNADO chart ranking every variable by the spread between its highest and
# lowest treatment rate. This is the chart that answers the client's actual
# question - "what should we work on?" - and it is deliberately colour-coded by
# whether the employer can CONTROL the variable, which is the second half of
# that question.

# ------------------------------------------------------------------------------
# Compute the spread for each candidate driver, with a minimum-n guard so that
# a rate computed on a handful of respondents cannot top the ranking.
# ------------------------------------------------------------------------------
MIN_CELL = 20

CONTROLLABLE = {
    'care_options', 'benefits', 'seek_help', 'wellness_program',
    'anonymity', 'leave', 'mental_vs_physical',
}

drivers = ['work_interfere', 'family_history', 'care_options', 'benefits',
           'Gender_Clean', 'anonymity', 'obs_consequence', 'seek_help',
           'wellness_program', 'mental_vs_physical', 'leave', 'no_employees',
           'Age_Group', 'remote_work', 'tech_company', 'self_employed']

rows = []
for col in drivers:
    grp = df.groupby(col, observed=True)['treatment_binary'].agg(['mean', 'size'])
    grp = grp[grp['size'] >= MIN_CELL]            # discard unreliable cells
    if len(grp) < 2:
        continue
    rows.append({
        'variable'   : col,
        'low'        : grp['mean'].min() * 100,
        'high'       : grp['mean'].max() * 100,
        'spread'     : (grp['mean'].max() - grp['mean'].min()) * 100,
        'controllable': col in CONTROLLABLE,
    })

tor = pd.DataFrame(rows).sort_values('spread')

fig, ax = plt.subplots(figsize=(13, 7))

for i, r in enumerate(tor.itertuples()):
    colour = TEAL if r.controllable else GREY
    ax.plot([r.low, r.high], [i, i], color=colour, lw=7,
            solid_capstyle='round', alpha=0.85, zorder=2)
    ax.scatter([r.low, r.high], [i, i], s=60, color=colour,
               zorder=3, edgecolor='white', linewidth=1.2)
    ax.text(r.high + 1.6, i, f'{r.spread:.1f} pp', va='center',
            fontsize=10, fontweight='bold', color=INK)

ax.axvline(BASE_RATE, color=FLAME, ls=':', lw=2, zorder=1)
ax.text(BASE_RATE + 0.8, len(tor) - 0.4, f'base rate {BASE_RATE:.1f}%',
        fontsize=9, color=FLAME, rotation=90, va='top')

ax.set_yticks(range(len(tor)))
ax.set_yticklabels(tor['variable'].values, fontsize=10.5)
ax.set_xlim(5, 105)

legend_handles = [
    mpatches.Patch(color=TEAL, label='Employer CAN control'),
    mpatches.Patch(color=GREY, label='Employer cannot control'),
]
ax.legend(handles=legend_handles, fontsize=10, loc='lower right')

finish(ax, 'Every driver ranked by effect size',
       f'Bar spans lowest → highest treatment rate across a variable\'s levels '
       f'(cells with n < {MIN_CELL} excluded)',
       xlabel='% who sought treatment', ylabel='')

plt.tight_layout()
plt.show()

display(tor.sort_values('spread', ascending=False)
        [['variable', 'low', 'high', 'spread', 'controllable']]
        .round(1).reset_index(drop=True))
````

##### 1. Why did you pick the specific chart?

**Answer:** This is the chart that answers the client's actual question — *what should we work on?* — and the **tornado form** is chosen because it encodes the right quantity: each bar spans from a variable's lowest to its highest treatment rate, so the **bar length is the effect size** and the ranking is immediate.

The **colour coding by controllability is the second half of the answer**, and it is what makes this more than a ranked list. A variable's effect size only matters to a decision-maker in combination with whether they can do anything about it, so encoding both on one chart lets the reader find the top-ranked *green* bar — which is the actual recommendation — rather than the top-ranked bar overall.

The **minimum cell size of 20** is an essential guard: without it, a variable with one tiny extreme level would top the ranking on the strength of a handful of respondents. Excluding those cells makes the ranking reflect evidence rather than noise.

#### Chart - 21 - Correlation Heatmap

````python
# ==============================================================================
# Chart - 21 : Correlation Heatmap (Multivariate)
# ==============================================================================
# The frame is almost entirely categorical, so Pearson correlation cannot be
# applied directly. We build an ORDINALLY ENCODED copy, mapping each level to a
# number that respects its natural order. The encoding is explicit and
# documented - arbitrary label encoding would produce meaningless coefficients.
#
# Variables are then reordered by HIERARCHICAL CLUSTERING rather than left in
# input order, so that groups of mutually-correlated variables appear as visible
# blocks on the diagonal. That block structure is the actual finding.

from scipy.cluster.hierarchy import linkage, leaves_list
from scipy.spatial.distance import squareform

enc = pd.DataFrame(index=df.index)

# ---- Binary Yes/No ----------------------------------------------------------
for c in ['family_history', 'treatment', 'remote_work', 'tech_company',
          'self_employed', 'obs_consequence']:
    enc[c] = (df[c] == 'Yes').astype(int)

# ---- Three-level support variables -----------------------------------------
# 'Don't know' is coded 1, between No=0 and Yes=2. NOTE: Chart 16 showed it
# actually behaves at or BELOW 'No', so this encoding is a simplification -
# flagged here and discussed in the limitations below.
SUP_MAP = {'No': 0, "Don't know": 1, 'Not sure': 1, 'Yes': 2}
for c in ['benefits', 'care_options', 'wellness_program', 'seek_help',
          'anonymity', 'mental_vs_physical']:
    enc[c] = df[c].map(SUP_MAP)

# ---- Ordered severity / difficulty scales ----------------------------------
# .astype(str) first: these columns are pandas Categorical after wrangling, and
# mapping a Categorical returns another Categorical, which cannot be correlated.
enc['work_interfere'] = df['work_interfere'].astype(str).map(
    {'Not Applicable': 0, 'Never': 1, 'Rarely': 2, 'Sometimes': 3, 'Often': 4})
enc['leave_difficulty'] = df['leave'].astype(str).map(
    {'Very easy': 0, 'Somewhat easy': 1, "Don't know": 2,
     'Somewhat difficult': 3, 'Very difficult': 4})
enc['company_size'] = df['no_employees'].astype(str).map(
    {s: i for i, s in enumerate(SIZE_ORDER)})

# ---- Openness / stigma scales ----------------------------------------------
OPEN_MAP = {'No': 0, 'Some of them': 1, 'Maybe': 1, 'Yes': 2}
for c in ['coworkers', 'supervisor', 'mental_health_interview',
          'phys_health_interview', 'mental_health_consequence',
          'phys_health_consequence']:
    enc[c] = df[c].map(OPEN_MAP)

# ---- Numeric / engineered ---------------------------------------------------
enc['Age']           = df['Age']
enc['support_score'] = df['support_score']
enc['is_female']     = (df['Gender_Clean'] == 'Female').astype(int)

enc = enc.fillna(0).astype(float)

# ------------------------------------------------------------------------------
# Reorder by hierarchical clustering so correlated blocks sit together
# ------------------------------------------------------------------------------
corr = enc.corr()
# .to_numpy(copy=True) is required: the DataFrame's backing array is read-only,
# so fill_diagonal would raise on it directly.
dist = (1 - corr.abs()).to_numpy(copy=True)  # distance: highly correlated = close
np.fill_diagonal(dist, 0)
order = leaves_list(linkage(squareform(dist, checks=False),
                            method='average'))
corr_ord = corr.iloc[order, order]

fig, ax = plt.subplots(figsize=(14, 11.5))
mask = np.triu(np.ones_like(corr_ord, dtype=bool), k=1)

sns.heatmap(corr_ord, mask=mask, annot=True, fmt='.2f', cmap='vlag',
            center=0, vmin=-1, vmax=1, square=True,
            linewidths=0.7, linecolor='white',
            cbar_kws={'shrink': 0.6, 'label': 'Pearson correlation'},
            annot_kws={'size': 6.8}, ax=ax)
ax.set_title('Correlation matrix, ordered by hierarchical clustering',
             loc='left', fontsize=14, pad=18)
ax.text(0, -0.8, 'Blocks on the diagonal = groups of variables that move together',
        fontsize=10, color='#6B6B6B', transform=ax.transData)
plt.xticks(fontsize=8.5)
plt.yticks(fontsize=8.5, rotation=0)
plt.tight_layout()
plt.show()

# ---- Rank variables by correlation with the target -------------------------
tc = corr['treatment'].drop('treatment').sort_values(key=abs, ascending=False)
print('CORRELATION WITH treatment (ranked by absolute strength)')
print('=' * 58)
for var, val in tc.items():
    bar = '#' * int(abs(val) * 60)
    print(f'  {var:<27} {val:+.3f}  {bar}')
````

##### 1. Why did you pick the specific chart?

**Answer:** A heatmap is the standard tool for viewing **all pairwise relationships at once**, which is the only way to detect clustering, redundancy and multicollinearity — none of which is visible when variables are examined one pair at a time.

Three design decisions carry the weight here.

**Explicit ordinal encoding rather than arbitrary label encoding.** With 26 of 27 variables categorical, Pearson correlation cannot be applied to the raw frame. Calling `LabelEncoder` would assign numbers alphabetically and produce coefficients with no meaning. Each variable is instead mapped to a scale respecting its natural order, documented inline.

**Reordering by hierarchical clustering** is the choice that distinguishes this from a standard correlation plot. Left in input order, a correlation matrix is a wall of numbers; reordered so that mutually-correlated variables sit adjacent, **groups appear as visible blocks along the diagonal** — and that block structure is the actual finding, not a presentational nicety.

**Masking the upper triangle** removes redundant ink, since the matrix is symmetric.

##### 2. What is/are the insight(s) found from the chart?

**Answer:**

**Correlations with the target, ranked:** `work_interfere` strongest by a wide margin, then `family_history`, then **`care_options` as the strongest employer-controllable correlate**, then `is_female`, then `benefits`, `support_score`, `anonymity` and `obs_consequence` as modest but consistent positives. `Age`, `company_size`, `remote_work`, `tech_company` and `self_employed` all sit near zero — confirming the null results from Charts 17 and 18 through a completely independent method.

**The clustering reveals three structural findings invisible in the bivariate charts.**

**1. The support variables form a single tight block.** `benefits`, `care_options`, `wellness_program`, `seek_help` and `anonymity` correlate strongly **with each other**, not merely with the target. They are not five independent levers — they behave as five measurements of **one latent factor**, best described as *organisational openness about mental health*. A company that does one of these well tends to do all of them well.

**2. The stigma variables form a separate block.** `mental_health_consequence`, `mental_health_interview`, `coworkers` and `supervisor` cluster together and correlate only weakly with the support block. **Provision and psychological safety are structurally distinct dimensions** — which validates the Stage 2 / Stage 3 architecture used throughout this notebook, and means a company can score well on one while failing entirely on the other.

**3. The matched mental/physical pairs correlate with each other but sit at very different levels**, independently confirming that the Chart 12 gap is a genuine mental-health-specific penalty rather than general health privacy.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

**Answer:**

**Positive business impact.** The two-block structure is the most strategically useful output of the entire analysis, because it tells the client they have **two distinct problems requiring two distinct programmes** — and that solving one will not solve the other.

The support block is addressed through **provision and communication**: audit, document, publish, repeat. The stigma block is addressed through **culture and manager capability**: training, leadership modelling, visible non-retaliation, explicit confidentiality. A client funding only the first produces well-publicised benefits that frightened employees do not use; funding only the second produces an open culture with nothing concrete to offer. **This chart is the evidence that both are required.**

The tight clustering *within* the support block is also a practical planning insight: because those five variables move together, the client need not sequence five separate initiatives. They are manifestations of one organisational posture, so **a single coherent communication programme should move all of them at once** — which makes the business case cheaper than five line items would suggest.

The near-zero correlations for age, size and work arrangement provide **negative evidence of real value**, ruling out the structural explanations a sceptical stakeholder would otherwise raise.

**Risk of negative growth.** The methodological caveats here are substantial and must not be buried.

**Pearson correlation on ordinally encoded categorical data is an approximation, not a measurement.** It assumes equal spacing between adjacent levels — that "No" to "Don't know" equals "Don't know" to "Yes". **Chart 16 proved this false**, showing "Don't know" sitting at or *below* "No" on all four support variables. The encoding used here therefore actively misrepresents the shape of those relationships. Coefficients should be read as indicating **presence, direction and rough rank only — never as precise effect sizes**, and quoting one to three decimals in a board deck would convey false precision.

**All coefficients are modest in absolute terms.** None approaches the range where single-variable prediction becomes reliable. No individual's likelihood of needing support can be inferred from these variables.

**The multicollinearity within the support block is a genuine planning trap.** Because those five are so intercorrelated, **their individual bivariate effects are not additive** — the warning from Chart 20, now with a mechanism attached. A client forecasting the sum of the individual lifts will over-promise, and the credibility cost of that lands on the whole programme.

````python
# ==============================================================================
# Chart - 22 : Pair Plot (Multivariate)
# ==============================================================================
# A full pair plot over 25 encoded columns would be 600 panels and unreadable.
# We select the five variables the correlation analysis ranked most informative,
# which keeps the grid interpretable while retaining the substantive signal.
#
# hue='Treatment' is the analytical core: colouring by the target turns every
# panel into a visual test of CLASS SEPARABILITY.

PAIR_VARS = ['work_interfere', 'support_score', 'care_options',
             'leave_difficulty', 'Age']

pair_df = enc[PAIR_VARS].copy()
pair_df['Treatment'] = df['treatment'].values

# Jitter is essential here: the encoded variables take only a handful of integer
# values, so without it every point in a scatter panel lands on top of ~200
# others and the density is completely invisible.
rng = np.random.default_rng(42)
jittered = pair_df.copy()
for c in PAIR_VARS:
    if c != 'Age':
        jittered[c] = jittered[c] + rng.normal(0, 0.11, len(jittered))

g = sns.pairplot(
    jittered,
    vars=PAIR_VARS,
    hue='Treatment',
    palette={'Yes': TEAL, 'No': FLAME},
    diag_kind='kde',
    plot_kws={'alpha': 0.35, 's': 20, 'edgecolor': 'none'},
    diag_kws={'fill': True, 'alpha': 0.5, 'linewidth': 1.6},
    height=2.25,
    corner=True,                 # lower triangle only; the upper half is redundant
)
g.figure.suptitle('Pair plot of the five most informative variables, '
                  'coloured by treatment status',
                  x=0.02, ha='left', y=1.02, fontsize=13.5,
                  fontweight='bold', color=INK)
g.figure.text(0.02, 0.985,
              'Small jitter applied to ordinal variables so overlapping points '
              'are visible',
              fontsize=9.5, color='#6B6B6B')
plt.show()

print('Variables plotted:', PAIR_VARS)
print('\nGroup means by treatment status:')
display(pair_df.groupby('Treatment')[PAIR_VARS].mean().round(2))
````

##### 1. Why did you pick the specific chart?

**Answer:** A pair plot shows every pairwise scatter plus each variable's marginal distribution in one grid, making it the natural complement to the heatmap: the heatmap reduces each relationship to a single number, the pair plot shows the **shape** behind that number — non-linearity, clustering and class separation that a coefficient cannot express.

Four constraints keep it usable. **Only five variables** are plotted, selected from the heatmap ranking — the full 25-variable grid would be 600 panels. **`corner=True`** drops the redundant upper triangle. **`hue='Treatment'`** is the analytical core, turning every panel into a visual test of class separability.

And **jitter is essential rather than cosmetic.** The encoded variables take only a handful of integer values, so without it every point in a scatter panel lands atop roughly two hundred others and the density is completely invisible — a dense cluster and a sparse one would look identical. Adding small Gaussian noise to the ordinal variables restores the density information that the encoding destroyed.

##### 2. What is/are the insight(s) found from the chart?

**Answer:**

* **`work_interfere` shows by far the clearest separation on its diagonal.** The two KDE curves are visibly displaced — the treated group concentrating at higher severity codes, the untreated at lower. This is the visual signature of the dose-response relationship measured in Chart 14.
* **`support_score` and `care_options` show moderate separation**, with the treated group shifted upward. Consistent, but with heavy overlap.
* **`Age` shows essentially none** — the curves sit almost on top of one another, independently confirming Chart 17.
* **`leave_difficulty` shows mild separation in the counter-intuitive direction**, reflecting the reverse-causality pattern from Chart 19.

**The most important observation is what is absent: no pair of variables cleanly separates the two classes.** Every scatter panel shows teal and vermillion points thoroughly intermixed. There is **no two-dimensional region where treatment status can be read off with confidence.**

**Interpretation.** Treatment-seeking is **genuinely multifactorial** — the product of clinical need, personal history, cultural background, employer environment and individual disposition acting together, not of any one or two measurable factors. The overlap is not a failure of the analysis; it is an accurate depiction of how the phenomenon works.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

**Answer:**

**Positive business impact.** The absence of clean separation is, counter-intuitively, among the most useful findings in the project, because it forecloses a tempting and harmful strategy.

It means **no employee can be reliably flagged as "likely to need mental health support" from workplace variables.** Any client considering predictive identification of at-risk staff — a genuinely common impulse in People Analytics — now has direct visual evidence that the data does not support it. That saves wasted investment in a model that would not work, and avoids the far larger costs of deploying one that *appeared* to: false positives that stigmatise employees who are fine, false negatives creating a dangerous illusion of coverage, and the trust collapse when staff discover they are being scored.

The strategic conclusion is therefore **universal provision over targeted intervention.** Because need is distributed across the whole workforce and cannot be predicted from observable characteristics, support must be available to everyone by default, accessible without justification, and free of gatekeeping. This is simpler to administer than a targeted programme and does not require the organisation to hold sensitive inferences about individuals.

The `work_interfere` separation remains useful in the right form: as a **self-assessment prompt delivered to the individual** (per Chart 14), never as a manager-facing detection signal.

**Risk of negative growth.** The most serious risk is that a client sees the moderate `support_score` and `care_options` separation and **attempts individual prediction anyway.** The overlap is severe enough that such a model would perform barely above chance while carrying the full ethical and legal weight of profiling employees on inferred mental health status — data subject to special protection under **GDPR Article 9** and comparable regimes. This chart should be used explicitly as the argument against that project.

A second risk is **misreading the overlap as evidence that employer action does not matter.** It shows nothing of the kind. Chart 16 demonstrated a 27.7-point population-level difference by care-option awareness. **A factor can shift an entire population's behaviour substantially while remaining useless for classifying any individual within it** — and confusing population-level effect with individual-level predictability, in either direction, produces bad decisions: dismissing real levers as ineffective, or treating real levers as personal diagnostics.

Finally, even with jitter, **ordinal encoding produces visible banding** that a non-technical audience may mistake for structure in the data. Shown outside an analytical audience, the encoding must be explained — or the chart replaced with the heatmap and bivariate panels, which carry the same message without the artefact.

## **5. Solution to Business Objective**

#### What do you suggest the client to achieve Business Objective ?
Explain Briefly.

**Answer:**

The objective was to **raise the share of employees getting appropriate care and cut productivity loss from untreated conditions, at the lowest incremental spend.** The funnel analysis locates the leaks; the tornado chart (Chart 20) ranks the levers; the recommendations below are ordered by **cost-to-impact ratio**, and each one names the funnel stage it repairs.

---

### **Priority 1 — Audit provision, then publish it in plain language**
*Stage 2 (Awareness) · Effect size 27.7 pp · Cost: near zero*

Chart 16 established the finding this entire report rests on: **uncertainty performs at or below known absence on all four support variables.** Employees who did not know whether they had benefits sought treatment at **37.1%** — below those who knew they had none (48.2%). Not knowing is worse than knowing it's bad.

Chart 9 quantifies the addressable population: 32.4% don't know about benefits, 24.9% aren't sure of care options.

**Do this:** audit what is actually provided *first*. Then publish a one-page plain-language summary — what's covered, how to access it, what it costs the employee, what the employer does and does not see. Surface it at onboarding, at annual enrolment, and in the manager toolkit.

**Sequence matters.** If the audit reveals the benefits genuinely are thin, an awareness campaign drives people toward support that doesn't exist — worse than silence.

---

### **Priority 2 — Verify anonymity systems, then guarantee it in writing**
*Stage 3 (Safety) · 65.1% uncertain — the largest single block in the dataset · Cost: near zero*

**Two-thirds of employees do not know whether using mental health resources stays confidential** (Chart 10). In a stigmatised decision under uncertainty, uncertainty resolves as "no" — and Chart 19 confirms it behaviourally: the uncertain group sits at 45.3%, **below even those who know anonymity is not protected** (57.8%).

**Do this:** audit the data pipeline — claims reporting, insurance records, approval workflows — and verify that usage genuinely cannot reach a manager. Fix what can. **Only then** publish a specific written guarantee naming who sees what.

**Order is non-negotiable.** Trust in confidentiality is asymmetric: built slowly across many non-events, destroyed permanently by one breach — and that story travels faster than any campaign.

---

### **Priority 3 — Train managers, not the whole organisation**
*Stage 3 (Safety) · Targeted and scalable*

Employees are **more than twice as likely to be fully open with their supervisor (41.0%) than with coworkers (17.9%)**. Broad cultural campaigning aims at the wrong target.

Manager training is also tractable: a 500-person company has perhaps 60 managers. That is a budgetable project with a completion state, unlike shifting the ambient culture of 500 people.

**Do this:** train managers on responding to disclosure, available accommodations, what must stay confidential, and which well-meaning responses cause harm. **Pair it with real authority to grant flexibility** — training without authority manufactures cynicism, which is worse than no training. Measure at team level, since the 31.2% who won't talk to their manager are mostly describing a specific person.

---

### **Priority 4 — Intervene at the recognition threshold**
*Stage 1 (Need) → Stage 4 (Action) · The highest-leverage timing*

Chart 14 revealed that help-seeking is a **step function, not a gradient**: the jump from "Never" to "Rarely" interference is **+56.3 points**, while the entire climb from "Rarely" to "Often" adds only +14.5. People act when they first *notice*, not when it gets bad.

The severe end is already seeking treatment at 85% — marginal return on reaching them further is low. **The opportunity is upstream.**

**Do this:** deploy brief self-assessment tools helping people recognise early impact themselves; write plain-language descriptions of early symptoms; train managers to ask about workload and focus rather than about mental health directly.

**Critically: self-directed, never manager-led detection.** Training managers to watch for signs of struggle is surveillance, and the rational employee response is to conceal better — pushing the threshold *later*, the opposite of the intent.

---

### **Priority 5 — Design for the group least likely to self-refer**
*Largest absolute block of unmet need*

Men are **79% of this workforce and seek treatment at 45.5% versus 68.8% for women** (Chart 15). Read correctly this is a **help-seeking gap, not a prevalence gap** — making untreated men the largest block of unmet need in the dataset.

**Do this:** frame support around performance, sleep, focus and stress rather than therapeutic language; deliver through low-disclosure, opt-out-by-default channels; use visible senior male role models, one of the few interventions the literature finds moves male help-seeking.

Separately, target **normalisation content at first-timers** — the 64.6% of the no-family-history group who have never sought treatment lack any personal template for what therapy involves (Chart 13).

---

### **What to expect — and what to avoid**

**Budget for utilisation to rise.** If this works, claims, EAP usage and leave requests all increase. Brief leadership *before* launch and frame the rise as the success metric — otherwise a working programme gets cut at the first budget review for the crime of working.

**Never attempt predictive targeting.** Chart 22 showed no pair of variables cleanly separates the classes. Individual risk scoring would perform barely above chance while carrying the full weight of profiling employees on inferred mental health status — protected under GDPR Article 9 and comparable regimes. **Universal provision, not targeted intervention.**

**Do not sum the effect sizes.** Chart 21 showed the five support variables are one latent factor measured five ways. Forecasting 27.7 + 26.7 + 15.5 would over-promise badly, and the credibility cost lands on the whole programme.

**Never solicit disclosure.** Chart 12's stigma gap invites a legally hazardous response. Pre-offer health enquiries are restricted under the ADA and the UK Equality Act. Make the environment safe and state policies proactively — do not ask.

**Do not export globally without local research.** 80% of the sample is US/UK/Canada; India contributes 10 respondents. Disclosure norms vary enormously, and a campaign that lands in Seattle can read as an invitation to self-incriminate elsewhere. **Treat this as a Western baseline.**

# **Conclusion**

This project analysed 1,259 responses to the 2014 OSMI Mental Health in Tech Survey, structured around a single question: **where between "I need help" and "I got help" do people actually drop out?**

**The funnel gave an unexpected answer.** Need is enormous — **79.0%** report a condition, and **62.1%** say it interferes with their work at least sometimes. But awareness collapses immediately after: only **35.3%** can name their care options, and only **29.8%** are confident their anonymity is protected. The largest single leak — over 43 percentage points — sits between Need and Awareness, which is the cheapest stage in the entire funnel to repair.

**The most revealing structural finding was that Action (50.6%) exceeds both Awareness and Safety.** More than half sought treatment despite not knowing their options and despite not trusting confidentiality. They went around the employer entirely — private care, their own insurance, a GP, out of pocket. **The employer pathway is not the main route to treatment for most of these people.** The company is getting little leverage and little credit from provision it may already be funding.

**Uncertainty is not a soft middle ground — it is the worst state to be in.** This replicated across four independent variables in Chart 16 and again on anonymity in Chart 19. Employees who did not know whether they had benefits sought treatment at **37.1%**, *below* those who knew they had none (**48.2%**). Someone who knows their employer offers nothing can arrange private care; someone in uncertainty is paralysed, because resolving the uncertainty itself requires a disclosure they are unwilling to make. **From a behavioural standpoint, an unknown benefit is worse than no benefit.**

**The controllable levers are nearly as powerful as the uncontrollable ones.** The tornado chart ranked every driver by effect size and colour-coded it by whether an employer can act on it. Symptom severity (~70.8 pp) and family history (38.6 pp) dominate — and neither is a lever. But **care-option awareness delivers 27.7 pp, roughly 72% of family history's effect, and it is fixable with a document rather than a budget.** That ratio is the commercial heart of this project.

**Structure turned out not to matter.** Company size, remote work, industry, employment status and age all sat inside the "no meaningful difference" band. A five-person company posted the highest treatment rate in the dataset, above companies with more than a thousand employees. **Good outcomes do not require scale, and scale does not produce them.**

**The stigma is real, specific, and measurable.** Because the survey asked matched mental and physical versions of the same questions, the mental-health-specific penalty could be isolated cleanly from ordinary health privacy. Employees are **4.8× more likely to fear consequences** from disclosing a mental health issue, and **4.6× less likely to raise it in an interview** — only **3.5%** would, against 16.0% for a physical condition. Meanwhile **14.6% have personally witnessed a colleague penalised**, which suggests the fear is not irrational.

**The multivariate analysis revealed two problems rather than one.** Hierarchical clustering showed the support variables forming one tight block — a single latent "organisational openness" factor — and the stigma variables forming a separate, only weakly related block. **Provision and psychological safety are structurally independent**, which means two programmes are required and neither alone will work: publicised benefits that frightened employees won't touch, or an open culture with nothing concrete to offer.

**The pair plot delivered a valuable negative result.** No pair of variables cleanly separates those who sought treatment from those who didn't. Treatment-seeking is genuinely multifactorial, so **individual-level prediction is not supportable** — making universal provision not merely the kinder strategy but the only defensible one, and simultaneously cheaper to administer and safer legally.

**Limitations, stated plainly.** The sample is 80% US/UK/Canada (India: 10 respondents), 79% male, and three-quarters aged 26–36. It is self-reported with no clinical validation, and self-selection likely inflates the 50.6% treatment figure. **Every relationship here is correlational**, and several are demonstrably confounded: `leave` read literally would suggest that making leave *harder* increases treatment-seeking, which is absurd and is pure reverse causation. The ordinal encoding underpinning Chart 21 assumes equal spacing between levels — an assumption Chart 16 proved false for exactly the variables that matter most. And the data is now over a decade old, predating both mass remote work and a substantial shift in public discourse.

**What survives all of that is a clear, cheap course of action.** Audit provision, then publish it in plain language. Verify the anonymity systems, then guarantee confidentiality in writing. Train the sixty managers rather than campaigning at the five hundred employees, and give those managers real authority. Help people recognise their own threshold early, through self-assessment and never through manager-led detection. Design outreach for the group least likely to self-refer. **None of it requires a new vendor or significant budget. Every item addresses a gap this data measured directly.**

The defensible one-line summary: **the tech industry's mental health problem is, to a substantial degree, a communication and trust problem sitting on top of a provision problem — and the communication half is by far the cheaper to fix.**
